```python
import pandas as pd
```

```python
df = pd.read_csv('name_zip_fips_state.csv')
```

```python
name_list = [x.split(' ') for x in df['name']]
first_name_list = [x[0] for x in name_list]
first_name_list
last_name_list = []
for name in name_list:
    i = -1
    last_name_found = False
    while last_name_found == False:
        if '.' in name[i]:
            i -= 1
        else:
            last_name_list.append(name[i])
            last_name_found = True
last_name_list
```

```python
df.insert(1, 'first name', first_name_list)
df.insert(1, 'last name', last_name_list)
```

```python
df.to_csv('name_zip_fips_state.csv')
```
