```python
import pandas as pd
import string
```

```python
df = pd.read_csv('name_zip_fips_state_Dec_18_with_missing.csv', dtype = str)
#Add leading zeros to the dataframe for the state and fips columns
df.astype({'fips': 'str'}).dtypes
df['fips'] = df['fips'].str.zfill(3)
df
```

```python
name_list = df['name']
last_name_list = []
name_extentions_list = ['jr', 'i', 'ii', 'iii', 'iv', 'v', 'vi', 'sr', 'jd', 'rn', 'bsn', 'cpa', 'cgma', 'ctp', 'mba', 'cma', 'pe', 'esq', '-']
for name in name_list:
    while ',' in name:
        name = name.split(',')[0]
    while ' (' in name:
        name = name.split(' (')[0]
    name = name.split(' ')
    i = -1
    last_name_found = False
    while last_name_found == False:
        if any(name_suf == name[i].lower() for name_suf in name_extentions_list):
            i -= 1
        elif '.' in name[i]:
            i -= 1
        else:
            last_name_list.append(name[i])
            last_name_found = True
last_name_list
```

```python
name_list = [x.split(' ') for x in df['name']]
full_name_last_name_zip = list(zip(name_list, last_name_list))
full_name_last_name_zip
```

```python
first_name_list = []
title_list = ['mrs', 'hon']
for x in full_name_last_name_zip:
    first_name_found = False
    sub_name_index = 0
    #Find the number of sub-names in the name
    subnames_len = len(x[0])
    #Remove punctuation:
    while sub_name_index < subnames_len:
        #Get the current clean name
        clean_name = x[0][sub_name_index].translate(str.maketrans('', '', string.punctuation))
        if clean_name == x[1]:
            break
        #If the name is less than 2 characters, move to the next name
        if len(clean_name) <= 2 or clean_name.lower() in title_list:
            sub_name_index += 1
        else:
            first_name_found = True
            break
    if first_name_found:
        first_name_list.append(clean_name)
    else:
        first_name_list.append('')
first_name_list
```

```python
len(first_name_list)
```

```python
df.insert(1, 'first name', first_name_list)
df.insert(1, 'surname', last_name_list)
df['state'] = df['state'].apply(lambda x: x.lower())
df.drop(df[df['state'] == 'no data'].index, inplace = True)
df = df.rename(columns={"fips": "county"})
```

```python
df['county'] = df['county'].astype(str)
col_list = ['surname', 'first name', 'name', 'state', 'county']
df = df[col_list]
```

```python
df.to_csv('Dec_18_name_zip_fips_state_with_names.csv')
len(df)
```
